```python
import os, requests
from ddgs import DDGS

out_dir = "/content/pilaf_images"
os.makedirs(out_dir, exist_ok=True)

queries = ["uzbek plov", "plov rice dish", "osh plov uzbekistan"]
headers = {"User-Agent": "Mozilla/5.0"}
seen = set()
count = 0

for q in queries:
    results = DDGS().images(q, max_results=60)
    for r in results:
        url = r["image"]
        if url in seen:
            continue
        seen.add(url)
        try:
            resp = requests.get(url, headers=headers, timeout=10)
            if resp.status_code == 200 and resp.headers.get("Content-Type", "").startswith("image"):
                with open(f"{out_dir}/plov_{count:04d}.jpg", "wb") as f:
                    f.write(resp.content)
                count += 1
        except Exception:
            pass  # skip sites that block or time out

print(count, "images downloaded")
```

```python
import os, requests
from ddgs import DDGS

def download_images(queries, out_dir, prefix, per_query=60):
    os.makedirs(out_dir, exist_ok=True)
    headers = {"User-Agent": "Mozilla/5.0"}
    seen = set()
    count = 0
    for q in queries:
        for r in DDGS().images(q, max_results=per_query):
            url = r["image"]
            if url in seen:
                continue
            seen.add(url)
            try:
                resp = requests.get(url, headers=headers, timeout=10)
                if resp.status_code == 200 and resp.headers.get("Content-Type", "").startswith("image"):
                    with open(f"{out_dir}/{prefix}_{count:04d}.jpg", "wb") as f:
                        f.write(resp.content)
                    count += 1
            except Exception:
                pass
    print(count, f"images downloaded to {out_dir}")

download_images(
    ["shashlik skewers", "uzbek shashlik", "lamb shashlik grill"],
    "/content/kebab_images",
    "kebab"
)
```
